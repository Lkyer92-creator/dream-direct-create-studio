# 🎤 Free Song Transcriber (Whisper)

Upload a song → get timestamped lyrics out. Runs the open-source Whisper model on a free Colab GPU — no API key, no cost.

**Before running:** in Colab go to **Runtime → Change runtime type → Hardware accelerator → T4 GPU** → Save.

Run the cells top to bottom. Works on any song — yours, covers, anything with vocals.

```
# 1. Install Whisper (open-source, free)
!pip install -q openai-whisper
```

```
# 2. Upload your song (MP3, WAV, M4A…)
from google.colab import files

uploaded = files.upload()
audio_path = list(uploaded.keys())[0]
print("Uploaded:", audio_path)
```

```
# 3. Transcribe — a full song takes ~1-2 minutes on the free GPU
import whisper

# "large-v3" = most accurate (best for sung vocals). Use "medium" if you want it faster.
model = whisper.load_model("large-v3")
print("Model loaded ✓  Transcribing…")
result = model.transcribe(audio_path, verbose=False)
print("Done ✓")
```

```
# 4. Show timestamped lyrics + save files to download
def ts(s):
    m, s = divmod(int(s), 60)
    return f"[{m:02d}:{s:02d}]"

timestamped = "\n".join(f"{ts(seg['start'])} {seg['text'].strip()}" for seg in result["segments"])
print(timestamped)
print()

# Clean version (lyrics only) — paste straight into DDC Studio / Lyric Lab
lyrics_only = "\n".join(seg["text"].strip() for seg in result["segments"])
with open("lyrics.txt", "w") as f:
    f.write(lyrics_only + "\n")
with open("lyrics_timestamped.txt", "w") as f:
    f.write(timestamped + "\n")

print("Downloading both files…")
files.download("lyrics.txt")
files.download("lyrics_timestamped.txt")
```
